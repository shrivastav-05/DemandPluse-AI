```python
import pandas as pd

df = pd.read_csv("Kaggle_Ecommerce Data.csv")

df = df.dropna().drop_duplicates()

df = df.drop(columns=[
    "order_id",
    "customer_id",
    "product_id",
    "order_date",
    "delivered_date",
    "request_date",
    "return_reason"
])

df["returned"] = df["returned"].map({"Yes": 1, "No": 0})

df = pd.get_dummies(
    df,
    columns=["category", "payment_method", "region", "customer_gender"],
    dtype=int
)

X = df.drop("returned", axis=1)
y = df["returned"]

print(X.select_dtypes(include="object").columns)
```
