```python
df["Engine_CC"] = df["Engine"].str.split().str[0].astype(float)
df["Engine_CC"] = df["Engine_CC"].fillna(df["Engine_CC"].mean())
```

```python
df["Power_bhp"] = ( df["Power"].str.split().str[0].replace("null", np.nan).astype(float))
df["Power_bhp"] = df["Power_bhp"].fillna(df["Power_bhp"].mean())
```

```python
df.loc[df["Seats"]<5,"Seats"]=5
#df["Seats"] = df["Seats"].fillna().mode()
df["Seats"] = df["Seats"].fillna(df["Seats"].mode())
```
